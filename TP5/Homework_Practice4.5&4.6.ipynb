```python
class Temperature:
    
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return ( 9 / 5 ) * celsius + 32
    
    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return ( 5 / 9 ) * (fahrenheit - 32)
    
    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15
    
org_celsius = 25

fahrenheit = Temperature.celsius_to_fahrenheit(org_celsius)
convert_celsisu = Temperature.fahrenheit_to_celsius(fahrenheit)
kelvin = Temperature.celsius_to_kelvin(org_celsius)


print(f"1 Celsius : {org_celsius}")
print(f"2 Fahrenheit : {fahrenheit}")
print(f"3 Kelvin : {kelvin}")
```

```python
class Product:
    store_name = "ITC Online Store"
    product_count = 0

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

        Product.product_count += 1
        self.product_number = Product.product_count

    def display_product(self):
        print(f"Product {self.product_number}")
        print(f"Name : {self.name}")
        print(f"Price : ${self.price}")
        print(f"Quantity : {self.quantity}")
        print(f"Total Value : ${self.total_value()}")
        print()

    def total_value(self):
        if Product.is_valid_price(self.price):
            return self.price * self.quantity
        else:
            return 0

    @classmethod
    def get_product_count(cls):
        return cls.product_count

    @staticmethod
    def is_valid_price(price):
        if price <= 0:
            return False
        return True


p1 = Product("Laptop", 800, 5)
p2 = Product("Mouse", 15, 20)
p3 = Product("Keyboard", 30, 10)

print("========== ITC ONLINE STORE ==========")
print()

p1.display_product()
p2.display_product()
p3.display_product()

total_inventory_value = (
    p1.total_value() + p2.total_value() + p3.total_value()
)

print(f"Total Products : {Product.get_product_count()}")
print(f"Total Inventory Value : ${total_inventory_value}")
```

## Questions and Answers

### 1. What is the difference between instance attributes and class attributes?

- **Instance attributes** belong to individual objects. Each object can have its own values, such as `name`, `price`, and `quantity`.
- **Class attributes** belong to the class and are shared by all objects, such as `store_name` and `product_count`.

### 2. Why should we use a class method to count products?

Because `product_count` is a class attribute. A class method uses `cls` to access and manage class attributes, allowing us to get the total number of products created.

### 3. Why is `is_valid_price()` implemented as a static method?

Because it only needs the `price` parameter to check whether the price is positive. It does not need to access instance attributes or class attributes.

### 4. Which methods in this exercise use `self`, `cls`, or neither?

| Method | Uses | Reason |
|---|---|---|
| `__init__()` | `self` | Initializes object attributes. |
| `display_product()` | `self` | Displays information about an object. |
| `total_value()` | `self` | Calculates the total value of a product. |
| `get_product_count()` | `cls` | Accesses the class attribute `product_count`. |
| `is_valid_price()` | Neither | Only checks the given price. |
