### Exercise 2: Rectangle

```python
class Rectangle:
    """Represent a rectangle with width and length."""

    def __init__(self, width, length):
        self.width = width
        self.length = length

    def perimeter(self):
        return 2 * (self.width + self.length)

    def area(self):
        return self.width * self.length

    def display_result(self):
        """Display the perimeter and area of the rectangle."""
        print(f"Perimeter: {self.perimeter()}")
        print(f"Area: {self.area()}")


while True:
    try:
        width = float(input("Enter width: "))

        if width < 0:
            print("INVALID: Width cannot be negative.")
            break

        length = float(input("Enter length: "))

        if length < 0:
            print("INVALID: Length cannot be negative.")
        elif width > length:
            print("INVALID: Width cannot be greater than length.")
        else:
            rectangle = Rectangle(width, length)
            rectangle.display_result()
            break

    except ValueError:
        print("INVALID: Please enter a numeric value.")
        break
```

### Exercise 3: Equation Calculator

```python
class EquationCalculator:
    def __init__(self, y, z):
        self.y = y
        self.z = z

    def calculate_x(self):
        """"Validate inputs and calculate x based on the equation x = yz / (y + z)."""
        if self.y == 0:
            raise ValueError("y must not be 0.")
        if self.z == 0:
            raise ValueError("z must not be 0.")
        if self.y + self.z == 0:
            raise ValueError("y + z must not be 0.")

        #return value x = yz / (y + z)
        return (self.y * self.z) / (self.y + self.z)

    def display_result(self):
        """Show result of the calculation."""
        try:
            x = self.calculate_x()
            print(f"x = {x}")
        except ValueError as e:
            print(f"Error: {e}")

try:
    y = float(input("Enter y: "))
    z = float(input("Enter z: "))
    calculator = EquationCalculator(y, z)
    calculator.display_result()
except ValueError:
    print("Error: please enter valid numbers.")
```

### Exercise 4: Hundreds Counter

```python
class HundredsCounter:
    """Count complete hundreds."""
    
    def __init__(self, number):
        """Initialize the number."""
        self.number = number

    def count_hundreds(self):
        """Count complete hundreds."""
        return self.number // 100

try:
    number = int(input("Enter a non-negative integer: "))

    if number < 0:
        print("Error: Number must be non-negative.")
    else:
        print("Complete hundreds:", HundredsCounter(number).count_hundreds())

except ValueError:
    print("Error: Please enter a valid integer.")
```

### Exercise 5: Cheater Game

```python
class LuckyNumberGame:
    """A game that returns the input number plus one."""
    def __init__(self, number):
        self.number = number

    def get_next_number(self):
        """Return the next number."""
        return self.number + 1

    def display_result(self):
        """Display the result."""
        print("The next number is :", self.get_next_number())

# Get input from the user
try:
    num = int(input("Enter an integer : "))

    game = LuckyNumberGame(num)
    game.display_result()
except ValueError:
    print("Error: Please enter a valid integer.")
```
