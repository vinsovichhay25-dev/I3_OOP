### 2. Principles of Object-Oriented Programming (OOP)
Object-Oriented Programming (OOP) is based on four fundamental principles: Encapsulation, Abstraction, Inheritance, and Polymorphism. These principles improve code organization, reusability, flexibility, and maintainability.

#### 2.1 Encapsulation
__Definition__: Encapsulation combines data (attributes) and methods into a single unit (class) while controlling access to internal data.

__Key points__:
- Protects important data from direct modification.
- Provides controlled access through methods.
- Allows data validation.
- Improves code security and maintainability.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())
```

__Explanation__: The __balance attribute is private. Users interact with it through methods such as deposit() and get_balance() rather than modifying it directly.

#### 2.2 Abstraction
__Definition__: Abstraction hides complex implementation details and exposes only the essential features required by users.

__Key points__:
- Hides unnecessary implementation details.
- Reduces program complexity.
- Defines common structures through abstract classes.
- Improves flexibility and maintainability.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCard(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} by Credit Card")

class Cash(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} by Cash")

payment1 = CreditCard()
payment2 = Cash()

payment1.pay(100)
payment2.pay(50)
```

__Explanation__: The abstract class Payment defines a required method, pay(). Each subclass implements the payment operation differently.

#### 2.3 Inheritance
__Definition__: Inheritance allows a child class to acquire attributes and methods from a parent class, promoting code reusability.

__Key points__:
- Reuses existing code.
- Reduces code duplication.
- Allows child classes to extend existing functionality.
- Supports method overriding.

__Types of inheritance__: Single, Multiple, Multilevel, Hierarchical, and Hybrid.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Employee: {self.name}")

class Manager(Employee):
    def manage(self):
        print(f"{self.name} manages the team")

manager = Manager("John")
manager.display()
manager.manage()
```

__Explanation__: Manager inherits the attributes and methods of Employee. It also introduces its own method, manage(). The super() function can be used to access parent-class methods and constructors.

#### 2.4 Polymorphism
__Definition__: Polymorphism allows different objects to use the same method name while performing different behaviors.

__Key points__:
- Supports different behaviors through a common interface.
- Improves code flexibility.
- Reduces repetitive code.
- Allows different object types to be handled similarly.

```python
class Email:
    def send(self):
        print("Sending email notification")

class SMS:
    def send(self):
        print("Sending SMS notification")

notifications = [Email(), SMS()]

for notification in notifications:
    notification.send()
```

__Explanation__: Both classes implement the same method, send(), but produce different results depending on the object.
